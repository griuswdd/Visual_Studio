### value.counts() y .describe()

### Cargar las librerías

Importar la librería Pandas y definimos los datos

```python
import pandas as pd 

datos = {
    'Producto': ['Laptop', 'Teclado', 'Laptop', 'Monitor', 'Teclado', 'Mouse', 'Monitor', 'Laptop'],
    'Categoria': ['Electrónica', 'Accesorios', 'Electrónica', 'Electrónica', 'Accesorios', 'Accesorios', 'Electrónica', 'Electrónica'],
    'Region': ['Norte', 'Sur', 'Norte', 'Este', 'Sur', 'Norte', 'Este', 'Sur'],
    'Precio_Unitario': [1200, 45, 1100, 300, 50, 25, 320, 1250],
    'Cantidad': [2, 10, 1, 4, 12, 60, 2, 9],
    'Sueldo': [4500, 3200, 6000, 2800, 5200, 1900, 4100, 7500]
}
```

#### Creamos el data frame con los datos propuestos


```python
df = pd.DataFrame(datos)
```

#### Visualizamos las primeras 5 filas

```python
df.head()
```

#### Funcion .describe()
hace operaciones con cada conjunto de elementos que tiene solo números (int/float) como max, min, mean, etc.

```python


df.describe()



```

```python
df['Sueldo'].describe() #--> SOLO PARA UNA COLUMNA, en este caso la columna 'sueldo'
```

#### Funcion _value.counts()
número de coincidencias que hay en una columna de cada elemento

```python
df['Producto'].value_counts()
```
