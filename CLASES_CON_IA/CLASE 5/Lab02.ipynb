### Gráficas

Importar la librería Pandas y definir los datos


```python
import pandas as pd 

datos = {
    'Producto': ['Laptop', 'Teclado', 'Laptop', 'Monitor', 'Teclado', 'Mouse', 'Monitor', 'Laptop'],
    'Categoria': ['Electrónica', 'Accesorios', 'Electrónica', 'Electrónica', 'Accesorios', 'Accesorios', 'Electrónica', 'Electrónica'],
    'Region': ['Norte', 'Sur', 'Norte', 'Este', 'Sur', 'Norte', 'Este', 'Sur'],
    'Precio_Unitario': [1200, 45, 1100, 300, 50, 25, 320, 1250],
    'Cantidad': [2, 10, 1, 4, 12, 60, 2, 9],
    'Sueldo': [4500, 3200, 6000, 2800, 5200, 1900, 4100, 7500]
}

df = pd.DataFrame(datos)
```

Tipos de gráfico

-kind: 'bar' barras verticales
-kind: 'barh' barras horizontales
-kind: 'hist' Histograma
-kind: 'pie'

```python
df['Producto'].value_counts().plot(
    kind='bar',
    title='Cantidad de Ventas por Producto',
    xlabel='Producto',
    ylabel='Total de transcacciones',
    color='blue'

)
```

```python
df['Sueldo'].plot(
    kind='hist',
    title = 'distribución de sueldos',
    bins = 11,
    color = 'red'

)
```
