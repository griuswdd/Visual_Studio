### Ejercicio 3
Escribí una función resumen_categoria(df, categoria) que reciba el DataFrame y el nombre de una categoría, y devuelva una tupla con (cantidad_de_productos, total_facturado, precio_promedio) para esa categoría. Usá try/except para manejar el caso de que la categoría no exista, devolviendo (0, 0, 0).

```python
def resumen_categoria(df,categoria):
    try:
        cantidad_productos = df['Categoria'].value_counts()[categoria]   
    except KeyError:
        return (0,0,0)  

    total_facturado = df.groupby('Categoria')['Total'].sum()[categoria]
    precio_promedio =df.groupby('Categoria')['Total'].mean()[categoria]
    return (cantidad_productos,total_facturado,precio_promedio)

resumen_categoria(df,'oe que fue')
    

```

Creá una clase AnalizadorVentas que reciba un DataFrame en el __init__ y tenga tres métodos:

- producto_top() — devuelve el nombre del producto con mayor cantidad total vendida.
- region_top() — devuelve la región con mayor facturación total.
- resumen() — devuelve un diccionario con ambos resultados, más el total general facturado.

```python
class DataFrame:
    def __init__(self,data):
        self.dataframe = data
    def producto_top(self):
        a = self.dataframe.groupby('Producto')['Cantidad'].sum()
        maximo = a.max()

        for p in a.keys():
            
            if a[p] == maximo:
                return p
            
    def region_top(self):
        a = self.dataframe['Region'].value_counts()
        for k in a.keys():
            if a[k] == a.max():
                return k
        return a.keys()
    def resumen(self):
        return self.region_top()

data1 = DataFrame(df)
data1.resumen()
```
