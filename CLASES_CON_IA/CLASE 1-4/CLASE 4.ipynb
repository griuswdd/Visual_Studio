# NumPy

###  1.Creación y vectorización

```python
import numpy as np

arr1 = np.array([10,20,30,40])
arr2 = np.array([1,2,3,4])
```

#### Cálculos

```python
resultado = arr1 + 5
print(resultado)
```

```python
resultado = arr2 * arr1
print(resultado)
```

### 2.Indexación y filtrado booleano (Boolean Indexing)

```python
edades = np.array([10,67,20,15,18])
es_mayor = edades >= 18
print(es_mayor)
```

```python
mayores_edadd = edades[edades>= 18]
print(mayores_edadd)
```

### 3.Métodos Estadísticos Rápidos

```python
datos = np.array([10,15,20,25,100])

print(np.mean(datos))
print(np.median(datos))
print(np.std(datos))
print(np.sum(datos))
print(np.max(datos))
```
